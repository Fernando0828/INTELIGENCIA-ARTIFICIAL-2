## Evaluación de Modelos de Clasificación: Análisis y Métricas

### 1. El conjunto de datos MNIST

El dataset MNIST (*Modified National Institute of Standards and Technology*) es uno de los conjuntos de datos de referencia más utilizados en el aprendizaje automático y la visión por computador. Está compuesto por un gran volumen de imágenes de dígitos escritos a mano (del 0 al 9) normalizadas y centradas, con una resolución de 28x28 píxeles en escala de grises. Se emplea habitualmente como banco de pruebas estándar para evaluar y comparar algoritmos de clasificación, permitiendo medir la capacidad de un modelo para reconocer patrones visuales y generalizar ante variaciones en la escritura.

### 2. Clasificación

La clasificación es una tarea fundamental de aprendizaje supervisado cuyo objetivo es asignar una etiqueta de clase discreta a cada instancia de entrada. A partir de un conjunto de datos de entrenamiento caracterizado por atributos y etiquetas conocidas, el modelo aprende una función de decisión. En problemas como el de MNIST, nos enfrentamos a una clasificación multiclase (diez categorías posibles), donde el algoritmo debe procesar los datos de entrada y estimar la probabilidad correspondiente a cada dígito.

### 3. Matriz de confusión

La matriz de confusión es una herramienta tabular que permite visualizar de forma detallada el rendimiento de un modelo de clasificación al contraponer las predicciones realizadas frente a las etiquetas reales del conjunto de datos. Su estructura se compone de cuatro métricas clave en escenarios binarios (y se extiende de forma matricial para multiclase):

* **Verdaderos Positivos (TP):** Casos en los que el modelo predijo correctamente la clase positiva.
* **Falsos Positivos (FP):** Casos en los que el modelo predijo incorrectamente la clase positiva (error de tipo I).
* **Verdaderos Negativos (TN):** Casos en los que el modelo predijo correctamente la clase negativa.
* **Falsos Negativos (FN):** Casos en los que el modelo predijo incorrectamente la clase negativa (error de tipo II).

Esta representación permite identificar con precisión no solo el porcentaje de aciertos global, sino también los tipos específicos de confusiones o errores que comete el clasificador entre clases.

### 4. Precisión y Curva ROC

La precisión (*Precision*) cuantifica la calidad de las predicciones positivas del modelo, calculándose como la proporción de verdaderos positivos entre todas las instancias predichas como positivas: $\text{Precision} = \frac{TP}{TP + FP}$. Es una métrica crítica cuando el costo de un falso positivo es elevado.

Por otra parte, la curva ROC (*Receiver Operating Characteristic*) es una representación gráfica que evalúa el rendimiento de un modelo de clasificación binaria a lo largo de todos sus umbrales de decisión posibles. Grafica la tasa de verdaderos positivos (sensibilidad) frente a la tasa de falsos positivos (1 - especificidad). Para resumir esta curva en un único valor numérico se utiliza el AUC (*Area Under the Curve*), donde un valor cercano a 1.0 indica una capacidad de discriminación perfecta, mientras que un valor de 0.5 denota un rendimiento equivalente al azar.
