Algoritmo
1. Decision Tree Learning (Árboles de decisión): 	Es un modelo que representa decisiones mediante un árbol de preguntas y respuestas.	Clasifica o predice un resultado siguiendo diferentes condiciones.
---------------------------------------------------------------------------
2. Random Forest (Bosque aleatorio): Es un conjunto de varios árboles de decisión que trabajan juntos.	Combina las decisiones de muchos árboles para obtener predicciones más precisas.
--------------------------------------------------------------------------
3. Naive Bayes	Es un algoritmo de clasificación basado en probabilidades.	Calcula la probabilidad de que un dato pertenezca a una categoría.
---------------------------------------------------------------------------
4. k-Nearest Neighbors (kNN)	Es un algoritmo que compara un dato nuevo con los datos más cercanos.	Clasifica un dato según las características de sus vecinos más similares.
--------------------------------------------------------------------------
5. k-means	Es un algoritmo que agrupa datos similares en diferentes grupos.	Encuentra grupos o patrones dentro de los datos sin categorías previamente definidas.
---------------------------------------------------------------------------
6. Support Vector Machine (SVM)	Es un algoritmo que busca una frontera para separar diferentes categorías de datos.	Clasifica datos separándolos en diferentes grupos.
--------------------------------------------------------------------------
7. Stochastic Gradient Descent (SGD)	Es un método utilizado para ajustar y entrenar modelos de Machine Learning.	Reduce el error del modelo modificando sus parámetros poco a poco.
----------------------------------------------------------------------------
8. AdaBoost	Es un método que combina varios modelos sencillos para crear uno más fuerte.	Mejora las predicciones enfocándose en los datos que fueron clasificados incorrectamente.
---------------------------------------------------------------------------
9. Q-learning	Es un algoritmo de aprendizaje por refuerzo basado en recompensas.	Aprende qué acciones tomar para conseguir la mayor recompensa posible.
---------------------------------------------------------------------------
10. Reinforcement Learning (Aprendizaje por refuerzo): Es un tipo de aprendizaje donde un agente aprende interactuando con un entorno.	Aprende mediante prueba, error, recompensas y penalizaciones.
