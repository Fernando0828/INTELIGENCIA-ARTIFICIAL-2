A continuación, se presenta el desarrollo de los cinco conceptos principales abordados en la
sesión, fundamentales para la construcción y optimización de modelos computacionales:
1. Bloques Fundamentales
Representan los cimientos sobre los cuales se construye cualquier arquitectura. En el
desarrollo de modelos, estos bloques incluyen las estructuras de datos (como los tensores)
y las capas básicas predefinidas (capas densas, convolucionales, recurrentes, etc.). Son las
unidades modulares esenciales que permiten estructurar el flujo de los datos.
2. Personalización Avanzada
Consiste en la capacidad de ir más allá de las configuraciones estándar. Permite estructurar
arquitecturas a medida, definir capas propias, funciones de pérdida personalizadas o
métricas específicas. Esta flexibilidad es crucial cuando se busca resolver problemas
particulares o implementar investigaciones novedosas que las herramientas
predeterminadas no logran abarcar.
3. Cálculo de Gradientes
Es el motor matemático que hace posible el aprendizaje. Mediante técnicas de
diferenciación automática, el sistema calcula los gradientes (las derivadas) de la función de
error con respecto a los parámetros del modelo. Esto permite determinar en qué dirección y
en qué magnitud deben ajustarse los pesos para minimizar las fallas, un proceso conocido
como retropropagación (backpropagation).
4. Bucles de Entrenamiento
Es el ciclo iterativo donde ocurre el aprendizaje real. En un bucle de entrenamiento
estándar, el modelo recibe un lote de datos, realiza una predicción, evalúa el margen de
error, calcula los gradientes y, finalmente, actualiza sus parámetros mediante un
optimizador. Este ciclo se repite de forma controlada a lo largo de varias épocas (iteraciones
sobre el conjunto de datos) hasta que el sistema alcanza un rendimiento óptimo.

5. Funciones y Grafos
Para maximizar la eficiencia computacional, las operaciones matemáticas complejas se
estructuran utilizando grafos de computación. Esto permite que los entornos de desarrollo
transformen el código dinámico en representaciones estáticas altamente optimizadas. El
uso de grafos acelera drásticamente el tiempo de procesamiento, paraleliza las tareas y
facilita la ejecución del modelo en hardware especializado, como las tarjetas gráficas
(GPUs).
