# Redes Neuronales y Keras

Este documento presenta una visión general y estructurada sobre los fundamentos de las redes neuronales artificiales y su implementación práctica utilizando las diferentes APIs de Keras.

---

### 01. De la Biología a las Redes

El concepto de las redes neuronales artificiales nace de la inspiración biológica en el funcionamiento de las neuronas del cerebro humano. Así como las neuronas biológicas reciben señales a través de las dendritas, las procesan y transmiten una respuesta mediante los axones, los modelos computacionales imitan este comportamiento matemático para reconocer patrones, aprender de los datos y resolver problemas complejos de aprendizaje automático.

### 02. El Perceptrón

El **perceptrón** es la unidad básica de una red neuronal y representa el modelo más sencillo de una neurona artificial. Consiste en:

* Una serie de entradas a las cuales se les asignan **pesos** específicos.


* Un **sesgo (bias)** que ayuda a desplazar la función de activación.


* Una función matemática que combina las entradas y determina si la neurona debe activarse o no (salida).

### 03. MLP y Retropropagación

* **MLP (Multilayer Perceptron / Perceptrón Multicapa):** Es una clase de red neuronal artificial feedforward que consta de al menos tres capas de nodos: una capa de entrada, una o más capas ocultas y una capa de salida. Cada capa está completamente conectada a la siguiente.


* **Retropropagación (Backpropagation):** Es el algoritmo fundamental para entrenar redes neuronales multicapa. Funciona calculando el error en la capa de salida y propagándolo hacia atrás a través de la red, ajustando los pesos de las conexiones mediante el descenso de gradiente para minimizar dicho error en iteraciones sucesivas.

### Implementación en Keras (APIs)

Keras ofrece diferentes enfoques y niveles de abstracción para construir y entrenar modelos de aprendizaje profundo según la flexibilidad que requiera el proyecto:

### 04. API Secuencial (`Sequential API`)

Es la forma más sencilla e intuitiva de construir modelos en Keras. Permite crear redes apilando las capas una tras otra de manera lineal (capa por capa). Es ideal para la gran mayoría de arquitecturas estándar donde la información fluye de manera secuencial desde la entrada hasta la salida.

### 05. API Funcional (`Functional API`)

Es un enfoque más flexible que la API Secuencial. Permite construir arquitecturas de redes neuronales más complejas, tales como:

* Modelos con múltiples entradas o múltiples salidas.


* Capas compartidas.


* Grafos acíclicos dirigidos (como conexiones residuales o arquitecturas tipo ResNet).



### 06. API de Subclases (`Subclassing API`)

Representa el enfoque más avanzado y flexible en Keras. Consiste en la programación orientada a objetos mediante la creación de clases personalizadas que heredan de `tf.keras.Model`. Permite definir explícitamente el método de forward pass (`call`), lo cual otorga un control total sobre la lógica del modelo, siendo muy útil para arquitecturas de investigación o dinámicas complejas.
