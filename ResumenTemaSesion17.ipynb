## Resumen sobre el Entrenamiento de Redes Profundas

A continuación, se presenta un resumen estructurado basado en los temas solicitados en la imagen:

* **Entrenamiento de redes profundas:** Proceso mediante el cual se ajustan los pesos y sesgos de una red neuronal con múltiples capas ocultas utilizando datos para minimizar el error de predicción.


* **Reto del gradiente e inestabilidad del gradiente:** Problemas comunes en redes profundas donde los gradientes se vuelven extremadamente pequeños (desvanecimiento) o extremadamente grandes (explosión), dificultando el aprendizaje en las capas iniciales.


* **Inicialización:** Técnicas para asignar los valores iniciales de los pesos (como Xavier o He) que evitan que la señal se degrade o sature en las primeras iteraciones.


* **Funciones de activación y funciones de activación modernas:** Elementos que introducen no linealidad en la red. Las opciones modernas (como ReLU, Leaky ReLU o GELU) ayudan a mitigar el problema del gradiente desvanecido.


* **Normalización:** Métodos como la normalización por lotes (*Batch Normalization*) que estabilizan y aceleran el entrenamiento al mantener las activaciones con media y varianza controladas.


* **Preentrenamiento y transferencia:** Estrategias que reutilizan modelos ya entrenados en tareas grandes (preentrenamiento) para aplicarlos a nuevos problemas específicos con menos datos (aprendizaje por transferencia).
